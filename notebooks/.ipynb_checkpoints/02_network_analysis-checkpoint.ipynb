# Network Analysis of Chronic Disease Comorbidities

```python
import pandas as pd
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt
```

```python
disease_df = pd.read_csv("../data/processed/processed_disease_matrix.csv")
disease_df.head()
```

```python
def compute_jaccard(df):
    cols = df.columns
    jaccard = pd.DataFrame(index=cols, columns=cols)
    
    for col1 in cols:
        for col2 in cols:
            intersection = ((df[col1] == 1) & (df[col2] == 1)).sum()
            union = ((df[col1] == 1) | (df[col2] == 1)).sum()
            jaccard.loc[col1, col2] = intersection / union if union != 0 else 0
            
    return jaccard.astype(float)

jaccard_matrix = compute_jaccard(disease_df)
jaccard_matrix
```

```python
jaccard_matrix['Diabetes_binary'].sort_values(ascending=False)
```

```python
G = nx.Graph()

# Add nodes
for col in disease_df.columns:
    G.add_node(col)

threshold = 0.20  # minimum similarity

for i in jaccard_matrix.index:
    for j in jaccard_matrix.columns:
        if i != j and jaccard_matrix.loc[i, j] > threshold:
            G.add_edge(i, j, weight=jaccard_matrix.loc[i, j])
```

```python
pos = nx.spring_layout(G, seed=42)

plt.figure(figsize=(10,8))
edges = G.edges(data=True)
weights = [edge[2]['weight']*10 for edge in edges]

nx.draw(G, pos, with_labels=True, width=weights, node_size=3000)
plt.title("Comorbidity Network (Jaccard Weighted)")
plt.show()
```

```python
degree_centrality = nx.degree_centrality(G)
betweenness_centrality = nx.betweenness_centrality(G)

centrality_df = pd.DataFrame({
    'DegreeCentrality': degree_centrality,
    'BetweennessCentrality': betweenness_centrality
}).sort_values(by='DegreeCentrality', ascending=False)

centrality_df
```
